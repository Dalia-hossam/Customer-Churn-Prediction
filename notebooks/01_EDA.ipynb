### 4. Relationship with Churn

- Tenure vs Churn 

> The relationship between customer tenure and churn is examined to determine whether customers with shorter or longer subscription are more likely to leave

```python
plt.figure(figsize=(8, 5))

ax = sns.boxplot(
    data=df,
    x="Churn",
    y="tenure"
)

ax.set_title("Customer Tenure by Churn Status", fontsize=14, fontweight="bold")
ax.set_xlabel("Churn Status")
ax.set_ylabel("Tenure (Months)")

plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()
```

```python
# compute and display skewness for numerical columns
skew_vals = df[numerical_cols].skew().round(4)
print(skew_vals)

# visualize skewness
plt.figure(figsize=(7,8))
skew_vals.plot(kind="bar", color="C1")
plt.axhline(0, color="k", linewidth=0.8)
plt.title("Skewness of Numerical Columns")
plt.ylabel("Skewness")
for i, v in enumerate(skew_vals):
    plt.text(i, v, str(v), ha="center", va="bottom" if v >= 0 else "top")
plt.tight_layout()
plt.show()
```

```python
df_copy = df.copy()

median_total = df_copy["TotalCharges"].median()
df_copy["TotalCharges_filled"] = df_copy["TotalCharges"].fillna(median_total)

skewness_before = df_copy["TotalCharges_filled"].skew()
df_copy["TotalCharges_transformed"] = np.log1p(df_copy["TotalCharges_filled"])
skewness_after = df_copy["TotalCharges_transformed"].skew()

print(f"Median TotalCharges used for filling: {median_total}")
print(f"Skewness before transformation: {skewness_before:.4f}")
print(f"Skewness after log1p transformation: {skewness_after:.4f}")

df_copy[["TotalCharges_filled", "TotalCharges_transformed"]].head()
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
sns.histplot(df_copy["TotalCharges_filled"], bins=30, kde=True, color="C0")
plt.title(f"TotalCharges (Filled) - Before Log1p\nSkewness: {skewness_before:.4f}")
plt.xlabel("TotalCharges")
plt.ylabel("Frequency")

plt.subplot(1, 2, 2)
sns.histplot(df_copy["TotalCharges_transformed"], bins=30, kde=True, color="C2")
plt.title(f"TotalCharges Transformed - After Log1p\nSkewness: {skewness_after:.4f}")
plt.xlabel("Log1p(TotalCharges)")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

## **Correlation Analysis**

This section examines the relationships among numerical features using a correlation Matrix

```python
plt.figure(figsize=(6, 5))

corr = df[numerical_cols].corr()

sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5,
    square=True,
    cbar_kws={"shrink": 0.8}
)

plt.title("Correlation Matrix of Numerical Features", fontsize=14, pad=12)
plt.xticks(rotation=0)
plt.yticks(rotation=0)
plt.tight_layout()

plt.show()
```

### Observation

- `TotalCharges` has a strong positive correlation with `tenure`, indicating that customers who stay longer accumulate higher total charges.
- `MonthlyCharges` has a weak correlation with both `tenure` and `TotalCharges`, suggesting that monthly pricing alone does not explain customer retention.
- No strong multicollinearity is observed among the numerical features, indicating that each variable provides unique information for further analysis.

## Key Business Insights

### Customer Retention

- Customer tenure is one of the strongest indicators of churn. Customers in the early stages of their subscription are considerably more likely to leave the company.
- Higher monthly charges are associated with an increased likelihood of churn, suggesting that pricing may influence customer retention.
- Customers with higher total charges generally remain with the company longer, reflecting greater customer loyalty and lifetime value.

### Customer Profile

- Gender has little influence on customer churn.
- Senior citizens show a moderately higher churn tendency than younger customers.
- Customers with partners and dependents are generally more loyal and less likely to leave the company.

### Service Performance

- Fiber Optic customers represent the highest-risk customer segment for churn and may require improvements in pricing, service quality, or customer support.
- Contract type is one of the strongest business drivers of churn. Customers on month-to-month contracts are substantially more likely to leave than those with one-year or two-year contracts.

### Value-Added Services

- Customers without Online Security and Tech Support experience significantly higher churn rates.
- Online Backup and Device Protection also contribute to improved customer retention by increasing customer engagement with the company's services.

### Billing & Payment Behavior

- Customers using Electronic Check have the highest churn rate among payment methods.
- Paperless Billing is associated with slightly higher churn, suggesting that this customer segment may benefit from targeted retention strategies.

---

## Business Recommendations

1. Prioritize retention campaigns for new customers during their first months of service.
2. Review pricing strategies and customer experience for Fiber Optic subscribers.
3. Encourage long-term contracts through discounts and loyalty incentives.
4. Promote bundled services that include Online Security, Tech Support, and Device Protection.
5. Offer personalized incentives to customers with high monthly charges who are identified as being at risk of churn.
6. Develop targeted retention campaigns for customers using Electronic Check and month-to-month contracts.
7. Leverage churn prediction models to proactively identify high-risk customers and intervene before cancellation.
