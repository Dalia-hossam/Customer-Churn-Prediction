## 7. **Exploratory Data Analysis (EDA)**   

> The purpose of this section is explore the dataset using statistical summaries and visualizations to better understand customer characteristics and identify factors that may influence customer churn

### 1. Churn distribution 
 Before analyzing the features, it is important to examine the distribution of the target variable (`Churn`) too understand the balane bettween customers who stayed and those who left

```python
df['Churn'].value_counts()
```

```python
df['Churn'].value_counts(normalize= True)*100
```

```python
plt.figure()
sns.countplot(data= df, x = "Churn")
plt.title("Customer Churn Distribution")
plt.xlabel("Churn")
plt.ylabel("Number of Customers")
plt.show()
```

### Observation 
 The target variable is moderately imbalanced. Approximately **73,45%** of customers did not churn, while **26,54** left the company. This indicates that most customers were retained, but the class imbalane should be considered during the model development and evaluation stage. 

### Numerical Features Analysis

The Numerical variables are analyzed to understand their distributions, detect potential outliers, and identifyy patterns that may be assosisted with customers churn

```python
numerical_cols = ["tenure", "MonthlyCharges", "TotalCharges"]

df[numerical_cols].describe().T
```

```python
for col in numerical_cols:
    plt.figure()
    sns.histplot(df[col], bins= 30, kde= True)
    plt.title(f"Distribution of{col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.show()
```

### 2. Distribution Analysis

The distribution of numerical features are visualized to identify skewness, concentration of values, and over all data patterns. 

### 3. Outlier Detection 

Boxplots are used to identify potential outliers in the numerical variables. 

```python
for col in numerical_cols: 
    plt.figure()
    sns.boxenplot(x=df[col])
    plt.title(f"Boxplot of {col}")
    plt.show()
```

### Observation 
 The boxplots do not reveal significant outliers in the numerial features. While `MonthlyCharges` and `TotalCharges` show slightly wider distribution, the observeed values appear to be valid customer records rather than data quality issues, therefore, no outlier treatment is required at this stage 

### 4. Relationship with Churn

- Tenure vs Churn 

> The relationship between customer tenure and churn is examined to determine whether customers with shorter or longer subscription are more likely to leave

```python
plt.figure()
sns.boxenplot(data= df, x = "Churn", y = "tenure")
plt.title("Tenure by Churn")
```

- Customers who churned generally have shorter tenures than those who remained. In contrast,  Customers with longer supscription periods are more likely to stay with the company, suggesting that customerrs loyalty increases over time

- Monthly charges vs Churn 

> This visualization exoplores whethr customers with higher monthly charges are more likely to churn

```python
plt.figure()
sns.boxenplot(data = df, x = "Churn", y = "MonthlyCharges")
plt.title("Monthly Charges by Churn")
plt.show()
```

- Customers who churned tend to have higher monthly charges compared to those who stayed. This suggests that higher monthly costs may be associated with an increased likelihood of customer churn

- Total Charges vs Churn 

> This analysis investigation the relationship between customers' total spending and churn behavior 

```python
plt.figure()
sns.boxenplot(data = df, x = "Churn", y = "TotalCharges")
plt.title("Total Chrges by Churn")
plt.show()
```

- Customers who stayed with the company generally have higher total charges than customers who churned. THis is expected becuse long-term customers accumulate higher total payments over time

## **7. Categorical Features Analysis**

### Categorical Feature vs churn

This section explores the relationship between categorical features and customer churn to identify customer characterstics associated with a higher likelihood of leaving the company

```python
categorical_features = df.select_dtypes(include=['object', 'str']).columns.to_list()

# Exclude identifier and target 

categorical_features.remove("customerID")
categorical_features.remove("Churn")
categorical_features
```

```python
for col in categorical_features:
    churn_rate =( pd.crosstab(df[col],df["Churn"],normalize= "index")*100)
    churn_rate.plot(kind="bar", figsize=(7,8))
    plt.title(f"Churn Rate by {col}")
    plt.ylabel("Precentage")
    plt.xticks(rotation = 20)
    plt.legend(title= "Churn")
    plt.tight_layout()
    plt.show()
```
